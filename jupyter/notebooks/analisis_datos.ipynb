# Análisis de Datos - Parcial 2 Comunicaciones
Este cuaderno consulta la base de datos PostgreSQL conectada en la red `backend_net`.

```python
# Instalación automática del conector de PostgreSQL
import sys
!{sys.executable} -m pip install -q psycopg2-binary

import psycopg2
import os

# Conexión a PostgreSQL dentro de la red interna de Docker
try:
    conn = psycopg2.connect(
        host=os.environ.get('POSTGRES_HOST', 'database'),
        port=os.environ.get('POSTGRES_PORT', 5432),
        dbname=os.environ.get('POSTGRES_DB', 'joomla_db'),
        user=os.environ.get('POSTGRES_USER', 'joomla_user'),
        password=os.environ.get('POSTGRES_PASSWORD', 'joomla_password')
    )
    cursor = conn.cursor()
    cursor.execute("SELECT table_name FROM information_schema.tables WHERE table_schema='public';")
    tablas = cursor.fetchall()
    print(f"Conexión exitosa. Se encontraron {len(tablas)} tablas en PostgreSQL:")
    for tabla in tablas[:10]:
        print(f" - {tabla[0]}")
    cursor.close()
    conn.close()
except Exception as e:
    print(f"Error de conexión: {e}")
```
